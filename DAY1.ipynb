```python
# --- KERANGKA KODE LATIHAN 1 ---
data_suhu = [30, 32, 35, 40, 28]

# 1. Tambahkan angka 45 ke dalam list data_suhu
# Tulis kode Anda di bawah ini:
data_suhu.append(45)

# 2. Cari suhu maksimum dan minimum
# Ganti tanda '...' dengan fungsi yang tepat (hint: max() dan min())
suhu_tertinggi = max(data_suhu)
suhu_terendah = min(data_suhu)

print("Data suhu sekarang:", data_suhu)
print("Suhu Tertinggi:", suhu_tertinggi)
print("Suhu Terendah:", suhu_terendah)

```

```python
# --- KERANGKA KODE LATIHAN 2 ---
model_config = {
    "nama_model": "XGBoost",
    "akurasi": 88.0,
    "status": "Eksperimen"
}

# 1. Ambil nilai dari key "status"
status_saat_ini = model_config["status"]

# 2. Ubah nilai "akurasi" menjadi 92.5
# Tulis kode Anda di bawah ini:
model_config["akurasi"] = 92.5

# 3. Tambahkan key baru "waktu_training" dengan nilai "2 jam"
# Tulis kode Anda di bawah ini:
model_config["waktu_training"] = "2 jam"

print("Status Model:", status_saat_ini)
print("Konfigurasi Terbaru:", model_config)
```

```python
# --- KERANGKA KODE LATIHAN 3 ---
umur_pengguna = [15, 22, 17, 30, 12, 19, 25]
kategori_dewasa = []

# Buat perulangan for dan logika if-else
for umur in umur_pengguna:
    # Tulis logika kondisi di bawah ini (jika umur >= 18, maka append ke kategori_dewasa)
    if umur >= 18:
        kategori_dewasa.append(umur)

print("Semua Umur:", umur_pengguna)
print("Pengguna Dewasa:", kategori_dewasa)
```
